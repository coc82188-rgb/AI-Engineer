# --------------------------------------------------
# 01. Pandas Basics
# --------------------------------------------------

```python
# Pandas is a python library mainly used for working with tabular data.
import pandas as pd     # importing pandas library and giving it an alias name 'pd'

# Series is basically a 1- dimensional array that can hold any data type.
marks = pd.Series([85, 90, 78, 92])     # pd.series(data)
print(marks)                            # printing the series

# we can also give our own labels to the series.
marks = pd.Series(
    [85, 90, 78],
    index=["Rahul", "Aman", "Priya"]
)
print(marks)

# accessing the data in series using index labels
print(marks["Rahul"])    # accessing data for Rahul
        
print(marks[0])         # accessing data for Rahul using index position
print(marks.mean())     # calculating the mean of the series
print(marks.max())      # calculating the maximum value of the series
print(marks.min())      # calculating the minimum value of the series
print(marks.describe())  # getting the statistical summary of the series, it includes count, mean, std, min, 25%, 50%, 75%, max values of the series
```

# --------------------------------------------------
# 02. Dataframe
# --------------------------------------------------

```python
# A DataFrame is a 2-dimentional array (table) made up of rows and columns. It is similar to a spreadsheet or SQL table.
data = {                                    
    "Name": ["Rahul", "Aman", "Priya"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}                                           # creating a dataframe using a dictionary where keys are column names and values are lists of column data
df = pd.DataFrame(data)                     # creating a dataframe using pd.DataFrame(data)
print(df)                                   # printing the dataframe

print(type(df))                             # printing the type of the dataframe

# selecting a column from the dataframe
print(df["Name"])                           # selecting the 'Name' column from the dataframe
print(df[["Name", "Marks"]])                # selecting multiple columns from the dataframe
print(df.loc[0])                            # selecting a row from the dataframe using index position
print(df.iloc[0])                           # selecting a row from the dataframe using index label
print(df.loc[0, "Name"])                     # selecting a specific value from the dataframe using index position and column name
```

# --------------------------------------------------
# 03. Loading real data with pandas
# --------------------------------------------------

```python
df = pd.read_csv("students.csv")        # reading a csv file and creating a dataframe from it
print(df)                               # printing the dataframe

df.head()                               # printing the first 5 rows of the dataframe
df.head(3)                              # printing the first 3 rows of the dataframe
df.tail()                               # printing the last 5 rows of the dataframe
df.tail(3)                              # printing the last 3 rows of the dataframe

# only the latest function will be executed, so if you want to see the output of all the above functions, you need to print them.

df.shape                                  # printing the shape of the dataframe (number of rows, number of columns)
df.columns                                # printing the column names of the dataframe
df.info()                                 # printing the information about the dataframe (number of non-null values, data types of columns)
df.describe()                             # printing the statistical summary of the dataframe (count, mean, std, min, 25%, 50%, 75%, max values of the numerical columns)
```

# --------------------------------------------------
# 04. Selecting Rows and Columns
# --------------------------------------------------

```python
df["Name"]                # select a comumn
df[["Name", "Marks"]]     # select multiple columns

# selecting using iloc
# iloc is used to select rows and columns by their integer index positions.

df.iloc[0]                # selecting the first row of the dataframe
df.iloc[0, 0]             # selecting the first row and first column of the
df.iloc[0:3]              # selecting the first 3 rows of the dataframe
df.iloc[0:3, 0:2]         # selecting the first 3 rows and first 2 columns of the dataframe
df.iloc[0:3, 2:4]         # selecting the first 3 rows and last 2 columns of the dataframe
# latest function will be executed, so if you want to see the output of all the above functions, you need to print them.

# select rows and columns with a condition
df[df["Marks"] > 80]      # selecting rows where the 'Marks' column
df[df["City"] == "Delhi"] # selecting rows where the 'City' column is equal to 'Delhi'
```

# --------------------------------------------------
# 05. Missing Data — Handling (NaN)
# --------------------------------------------------

```python
# Real world dataframe often have missing values
#| Name  | Age | Marks |
#| ----- | --: | ----: |
#| Rahul |  21 |    85 |
#| Aman  | NaN |    90 |
#| Priya |  20 |   NaN |

df_test = df.copy()             # creating a copy of the dataframe to demonstrate missing values
df_test.loc[1, "Age"] = None    # setting the value of the 'Age' column for the second row to None (missing value)
df_test.loc[2, "Marks"] = None  # setting the value of the 'Marks' column for the third row to None (missing value)
print(df_test)                  # displaying the dataframe with missing values

df_test.isnull()                # checking for missing values in the dataframe, it will return a dataframe of the same shape with True for missing values and False for non-missing values
df_test.isnull().sum()          # counting the number of missing values in each column of the dataframe


```

# --------------------------------------------------
# 06. dropna() — Removing Missing Data/ filna() - Filling Missing Data
# --------------------------------------------------

```python
# dropna() function is used to remove missing values from the dataframe.
clean_df = df_test.dropna()             # removing rows with missing values from the dataframe, it will return a new dataframe without missing values
print(clean_df)                         # displaying the dataframe after removing rows with missing values
# Aman → Age missing → row removed
# Priya → Marks missing → row removed

# fillna() function is used to fill missing values in the dataframe with a specified value.
df_test.fillna(0)                       # filling missing values in the dataframe with 0
print(df_test.fillna(0))                # displaying the dataframe after filling missing values with 0
```

# --------------------------------------------------
# 07. Sorting & Basic Aggregation
# --------------------------------------------------

```python
# sorting dataframe
print(df.sort_values("Marks"))                     # sorting the dataframe by the 'Marks' column in ascending order
print(df.sort_values("Marks", ascending=False))    # sorting the dataframe by the 'Marks' column in descending order

# value count()
print(df["City"].value_counts())                   # counting the number of occurrences of each unique value in the 'City' column of the dataframe

# basic operations on dataframe
print(df["Marks"].mean())                          # calculating the mean of the 'Marks' column in the dataframe
print(df["Marks"].sum())                           # calculating the sum of the 'Marks' column in the dataframe
print(df["Marks"].max())                           # calculating the maximum value of the 'Marks' column in the dataframe
print(df["Marks"].min())                           # calculating the minimum value of the 'Marks' column in the dataframe
print(df["Marks"].std())                           # calculating the standard deviation of the 'Marks' column in the dataframe
```

# --------------------------------------------------
# 8. Data Cleaning — Duplicates & Data Types
# --------------------------------------------------

```python
# checking for duplicates in the dataframe
print(df.duplicated())                             # checking for duplicate rows in the dataframe, it will return a boolean series
print(df.duplicated().sum())                       # counting the number of duplicate rows in the dataframe

# removing duplicates from the dataframe
df.drop_duplicates(inplace=True)                   # removing duplicate rows from the dataframe, it will modify the original dataframe in place

print(df.dtypes)                                   # printing the data types of each column in the dataframe
df["Marks"] = df["Marks"].astype(int)              # converting the data type of the 'Marks' column to integer
df["Age"] = df["Marks"].astype(float)                # converting the data type of the 'Marks' column to float

print(df["Marks"].dtype)

```
