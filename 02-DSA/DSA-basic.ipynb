```python
# --------------------------------------------------
# 01. Arrays / Lists
# --------------------------------------------------

numbers = [10, 20, 30, 40, 50]  #Initialize the array/list
print(numbers)

print(numbers[0])  #Access the element
print(numbers[2])  #Access the element at index 2 (3rd element )

numbers[2] = 100    #update the element at index 2
print(numbers)

for number in numbers:
    print(number)       #traverse the array/list
```

```python
# --------------------------------------------------
# 02. Linked Lists
# --------------------------------------------------

# A linked list stores data in nodes.
# Each node contains data and a reference to the next node.

class Node:                     # Initialize the Node class
    def __init__(self, data):   # Initialize the constructor with data parameter
        self.data = data        # Initialize the data attribute
        self.next = None        # Initialize the next pointer to None


# Create nodes
node1 = Node(10)    # Create a node with data 10
node2 = Node(20)    # Create a node with data 20
node3 = Node(30)    # Create a node with data 30
node4 = Node(40)

# Connect the nodes
node1.next = node2      #Connect node1 to node2
node2.next = node3      #Connect node2 to node3
node3.next = node4

# Traverse the linked list
current = node1     #Start from the head of the linked list

while current is not None:      #Traverse until the end of the linked list
    print(current.data)         #Print the data of the current node
    current = current.next      #Move to the next node
```

```python
# --------------------------------------------------
# 03. Stacks
# --------------------------------------------------

# Stack is a linear data structure that follows the Last In First Out (LIFO) principle.
# The last element added to the stack will be the first one to be removed.
# Basic operations of a stack are push (add an element to the top of the stack) and pop (remove the top element from the stack).

stack = []              #Initialize an empty stack, it is a list in Python

stack.append(10)        #Add an element to the top of the stack, (push operation)
stack.append(20)        #Add an element to the top of the stack, (push operation)
stack.append(30)        #Add an element to the top of the stack, (push operation)
print(stack)            #Print the stack

item = stack.pop()      #Remove the top element from the stack (pop operation)
print(item)             #Print the removed element
print(stack)            #Print the updated stack
```

```python
# --------------------------------------------------
# 04. Queues
# --------------------------------------------------

# Queue is a linear data structure that follows the First In First Out (FIFO) principle.
# The first element added to the queue will be the first one to be removed.
# Basic operations of a queue are enqueue (add an element to the end of the queue) and dequeue (remove the front element from the queue).


queue = []              #Initialize an empty queue, it is a list in Python

queue.append(10)        #Add an element to the end of the queue (enqueue operation)
queue.append(20)        #Add an element to the end of the queue (enqueue operation)
queue.append(30)        #Add an element to the end of the queue (enqueue operation)
print(queue)            #Print the queue

item = queue.pop(0)     #Remove the front element from the queue (dequeue operation)
print(item)             #Print the removed element
print(queue)            #Print the updated queue

```

```python
# --------------------------------------------------
# 05. Trees
# --------------------------------------------------

# Tree is a hierarchical data structure that consists of nodes, with a single node called the root, and zero or more child nodes for each parent node.
# The topmost node is called the root, and nodes with no children are called leaves, and nodes with children are called internal nodes.
 
# A tree node can be represented using a class in Python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Create nodes
root = Node(10)             #Create the root node with data 10
             
root.left = Node(20)        #Create a left child node with data 20
root.right = Node(30)       #Create a right child node with data 30

root.left.left = Node(40)   #Create a left child node of the left child with data 40 
root.left.right = Node(50)  #Create a right child node of the left child with data 50


#        10
#       /  \
#     20    30
#    /  \
#   40   50
```
