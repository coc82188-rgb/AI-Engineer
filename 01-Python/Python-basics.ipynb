```python
# --------------------------------------------------
# 13. Lists
# --------------------------------------------------
fruits = ["apple", "banana", "mango"]   #list_nmae = [item1, item2, item3, ...]
print(fruits)   #print the list
print(fruits[0])    #print the first item in the list
print(fruits[1])    #print the second item in the list
print(fruits[-1])    #print the last item in the list
print(fruits[1:3])   #print items from index 1 to 2 (3 is not included)
print(fruits[1:])   #print items from index 1 to end
print(len(fruits))   #print the number of items in the list

fruits.append("orange")  #add an item to the end of the list
fruits.remove("banana")  #remove an item from the list
fruits.insert(2, "grapes")   #insert an item at a specific index
print(fruits)  

fruits.sort()   #sort the list in ascending order
print(fruits)
```

```python
# --------------------------------------------------
# 14. Tuples
# --------------------------------------------------
numbers = (10, 20, 30, 40, 50)   #tuple_name = (item1, item2, item3, ...)
print(numbers)   #print the tuple
print(numbers[0])    #print the first item in the tuple
print(numbers[1])    #print the second item in the tuple
print(numbers[-1])    #print the last item in the tuple
print(numbers[1:3])   #print items from index 1 to 2 (3 is not included)
print(len(numbers))   #print the number of items in the tuple
print(type(numbers))   #print the type of the variable
print(numbers.count(20))   #count the number of occurrences of an item in the tuple
print(numbers.index(30))   #find the index of an item in the tuple
print(numbers + (60, 70))   #concatenate two tuples
print(numbers * 2)   #repeat the tuple
print(20 in numbers)   #check if an item is in the tuple

```

```python
# Note -  Difference between list and tuple is that list is mutable (can be changed) and tuple is immutable (cannot be changed)
# example of mutable and immutable
fruits = ["apple", "banana", "mango"]   # this is a list (mutable)
print(fruits)
fruits[0] = "orange"   # this is allowed
print(fruits)

numbers = (10, 20, 30, 40, 50)   # this is a tuple (immutable)
print(numbers)
# numbers[0] = 60   # this would raise an error
```

```python
# --------------------------------------------------
# 15. Dictionaries
# --------------------------------------------------
student = {
    "name": "Eklavya",  # key-value pair
    "age": 22,  
    "marks": 85 
}

print(student)
print(student["name"])
print(student["marks"])
student["city"] = "Ghaziabad" #add a new key-value pair to the dictionary
student["marks"] = 90 #update the value of an existing key in the dictionary
student.pop("age") #remove a key-value pair from the dictionary
print(student.keys()) #print all the keys in the dictionary
print(student.values()) #print all the values in the dictionary
print(student.items()) #print all the key-value pairs in the dictionary
print(len(student)) #print the number of key-value pairs in the dictionary
print(student)
```

```python
# --------------------------------------------------
# 16. Sets
# --------------------------------------------------
numbers = {10, 20, 30, 20, 10}  #set_name = {item1, item2, item3, ...} - set is a collection of unique items
# Note -  Set is unordered, unindexed and does not allow duplicate values
print(numbers)
numbers.add(40)     #add an item to the set at the start
print(numbers)
numbers.remove(20)  #remove an item from the set
print(numbers)
print(30 in numbers)    #check if an item is in the set
print(len(numbers))   #print the number of items in the set
print(numbers.union({50, 60}))   #combine two sets
print(numbers.intersection({30, 40, 50}))   #find the common items in two sets
```

FUNCTION WITH PARAMETER

```python
def greet(name):    #def function_name(parameter): - function definition with parameter
    print("Hello", name)    #print the value of the parameter

greet("Eklavya")    #function call with argument - calling the function with a value for the parameter
greet("Rahul")
```

FUNCTION WITH RETURN

```python
def add(a, b):  #def function_name(parameter1, parameter2): - function definition with parameters
    return a + b    #return statement - returns the value of the expression to the caller

result = add(10, 5)     #result variable stores the value returned by the function call
print(result)       #print the value of the result variable
print(result * 2)
print(result + 100)
```

numbers = [10, 25, 30, 7, 45, 12]

Create a program that

    Prints only the numbers greater than 20

    Counts how many numbers are greater than 20
    
    Calculates the total of those numbers

```python
numbers = [10, 25, 30, 7, 45, 12]

count = 0
total = 0

for number in numbers:
    if number > 20:
        print(number)
        total += number
        count += 1

print("Count of numbers greater than 20:", count)
print("Total of numbers greater than 20:", total)
```

QUESTION 4 - FUNCTION:

Create a function called calculate_average() that:

    Takes 3 marks as arguments
 
    Calculates their average
 
    Uses return to send the result back
    
    Stores the returned value in a variable
    
    Prints the average

```python
def calculate_average(a, b, c):
    average = sum([a, b, c]) / 3
    return average

average = calculate_average(80, 70, 90)

print(average)
```
